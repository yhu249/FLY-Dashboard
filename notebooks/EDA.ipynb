## Objective

This notebook performs exploratory data analysis (EDA) on the FLY Initiative datasets.

The goals are:

1. Understand the structure and meaning of each dataset.
2. Identify data quality issues.
3. Explore relationships between student participation and program outcomes.
4. Define requirements for the future data cleaning pipeline and stakeholder dashboard.

## Data Sources

The datasets include:

- Program Scores

## Outcome EDA


```python
# Summary statistics for assessment scores

assessment_cols = [
    "Total Score(Pre-assessment)",
    "Total Score(Post-assessment)"
]

scores[assessment_cols].describe()
```

```python
# Calculate score improvement for exploration

scores["Assessment Improvement"] = (
    scores["Total Score(Post-assessment)"]
    -
    scores["Total Score(Pre-assessment)"]
)

scores["Assessment Improvement"].describe()
```

```python
# Explore which modules show higher improvement

module_improvement = (
    scores
    .groupby("Assignment")["Assessment Improvement"]
    .agg(
        ["count", "mean", "median"]
    )
    .sort_values(
        "mean",
        ascending=False
    )
)

module_improvement
```

```python
# Attendance percentage distribution

scores["Attendance Percentage"].describe()
```

```python
# Relationship between attendance and learning improvement

scores[
    [
        "Attendance Percentage",
        "Assessment Improvement"
    ]
].corr()
```

- Pre- and post-assessment scores are available for 5,137 records, indicating sufficient data for evaluating student learning outcomes.

- The average total assessment score increased from 19.59 (pre-assessment) to 22.24 (post-assessment), with an average improvement of 2.65 points.

- Assessment improvement varies across financial literacy modules. Investing, Managing Credit, and Types of Credit show relatively higher average improvements compared with other modules.

- Attendance percentage has an average of 74.43%, with most students showing relatively high participation levels.

- Attendance percentage has a positive but weak correlation with assessment improvement (correlation coefficient = 0.17), suggesting that attendance may have some relationship with learning outcomes, but other factors may also influence improvement.

- Program Attendance contains 89,927 attendance records from 1,116 unique students.

- Attendance records are highly granular, with an average of approximately 80 attendance records per student. Therefore, attendance data should be aggregated before being joined with student outcome data.

- Attendance Status contains multiple categories:
    - No Class
    - P (Present)
    - A (Absent)
    - T
    - Pending Attendance

- Potential duplicate attendance records were identified (6,226 records) based on Student ID, Date, and Volunteer Job Name. Duplicate handling rules should consider the Created Date field, as newer records may represent updated attendance information.

- The attendance dataset covers dates from September 2025 to January 2026 based on the current data.
